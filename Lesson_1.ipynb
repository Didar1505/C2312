# Часть 1. Обычная программа

Сначала вспомним обычный Python.

# Часть 2. Дадим компьютеру примеры

Теперь не будем писать правило вроде:

```python
if rating >= 7:
```

Вместо этого покажем программе **готовые примеры**.

У каждого человека есть два признака:

1. сколько часов в неделю он играет;
2. какую оценку поставил игре.

Правильный ответ:

- `0` — игра не понравилась;
- `1` — игра понравилась.

````python
# X — входные данные, или ПРИЗНАКИ
# [часы игры в неделю, оценка игры]

X = [
    [1, 2],
    [2, 3],
    [2, 4],
    [3, 4],

    [5, 7],
    [6, 8],
    [7, 9],
    [8, 10]
]

# y — правильные ответы
# 0 = не понравилась
# 1 = понравилась

y = [
    0,
    0,
    0,
    0,
    1,
    1,
    1,
    1
]

print("Количество примеров:", len(X))
````

## Посмотрим на данные

Очень полезно не просто хранить числа, а понимать, что они означают.

Запустим цикл и красиво выведем все примеры.

````python
for i in range(len(X)):
    hours = X[i][0]
    rating = X[i][1]
    answer = y[i]

    print(
        f"Пример {i + 1}: "
        f"играет {hours} ч., "
        f"оценка {rating}, "
        f"ответ = {answer}"
    )
````

Пока можно не изменять данные окончательно: дальше мы используем исходный набор.

# Часть 3. Создаём нейронную сеть

Теперь создадим настоящий алгоритм нейронной сети.

`MLPClassifier` — это небольшая многослойная нейронная сеть.

Параметр:

```python
hidden_layer_sizes=(5,)
```

означает, что внутри будет скрытый слой из **5 нейронов**.

Схематично:

```text
часы  ───┐      ○
         ├──→   ○
оценка ──┘      ○  ───→ ответ
               ○
               ○
```

````python
model = MLPClassifier(
    hidden_layer_sizes=(5,),
    solver="lbfgs",
    max_iter=2000,
    random_state=42
)

print("Нейронная сеть создана!")
````

# Часть 4. Обучаем ИИ

Пока наша модель **ничего не знает**.

Теперь выполним самую важную команду:

```python
model.fit(X, y)
```

`fit()` означает:

> «Изучи примеры `X` и правильные ответы `y`, найди между ними закономерность».

````python
model.fit(X, y)

print("Обучение завершено! 🧠")
````

# Часть 5. Первые предсказания

Команда:

```python
model.predict(...)
```

просит уже обученную модель сделать предсказание.

Начнём с двух простых примеров.

````python
test_data = [
    [1, 2],
    [8, 9]
]

predictions = model.predict(test_data)

print("Предсказания:", predictions)
````

Ожидаем примерно:

```text
[0 1]
```

То есть:

- `[1, 2]` → скорее **не понравилось**;
- `[8, 9]` → скорее **понравилось**.

Но интереснее проверить данные, которых модель раньше не видела.

````python
new_people = [
    [4, 6],
    [5, 5],
    [10, 2]
]

predictions = model.predict(new_people)

for person, answer in zip(new_people, predictions):
    print(person, "→", answer)
````

### Перед следующим запуском попробуйте угадать сами

Что ответит ИИ для:

```text
[10, 2]
```

То есть человек:

- играет **10 часов**;
- поставил игре оценку **2**.

Что важнее для модели — количество часов или оценка?

Здесь уже становится интересно: мы не писали готовое правило, модель нашла его сама по примерам.

# Часть 6. Сделаем удобную функцию

Чтобы каждый раз не читать `0` и `1`, создадим функцию.

Она будет переводить ответ модели на обычный русский язык.

````python
def ask_ai(hours, rating):
    result = model.predict([[hours, rating]])[0]

    if result == 1:
        print("🤖 ИИ считает: игра понравилась")
    else:
        print("🤖 ИИ считает: игра не понравилась")
````

````python
ask_ai(7, 9)
ask_ai(2, 3)
ask_ai(4, 6)
````

# Часть 7. Ввод данных с клавиатуры

Теперь сделаем маленькую программу, с которой можно взаимодействовать.

# Часть 8. Насколько ИИ уверен?

Модель может не только выбрать `0` или `1`.

Она также может показать **примерную вероятность** каждого ответа.

Для этого есть:

```python
predict_proba()
```

````python
person = [[4, 6]]

probabilities = model.predict_proba(person)[0]

print("Вероятность ответа 0:", probabilities[0])
print("Вероятность ответа 1:", probabilities[1])
````

Сделаем вывод красивее — в процентах.

````python
person = [[4, 6]]

probabilities = model.predict_proba(person)[0]

print(f"Не понравилось: {probabilities[0] * 100:.1f}%")
print(f"Понравилось:    {probabilities[1] * 100:.1f}%")
````

### Задание

Проверьте вероятности для:

```python
[1, 2]
[8, 10]
[5, 5]
[10, 2]
```

Найдите вариант, где модель уверена меньше всего.

````python
tests = [
    [1, 2],
    [8, 10],
    [5, 5],
    [10, 2]
]

for person in tests:
    probabilities = model.predict_proba([person])[0]

    print(
        person,
        "→",
        f"0: {probabilities[0] * 100:.1f}%",
        f"1: {probabilities[1] * 100:.1f}%"
    )
````

# Часть 9. Увидим данные на графике

По вертикали — оценка игры.

````python
hours = [person[0] for person in X]
ratings = [person[1] for person in X]

plt.scatter(hours, ratings, c=y, s=100)

plt.xlabel("Часов игры в неделю")
plt.ylabel("Оценка игры")
plt.title("Данные, на которых учится ИИ")
plt.grid()

plt.show()
````

# Часть 10. Меняем данные → меняется ИИ

Это один из самых важных экспериментов занятия.

Сначала спросим исходную модель про человека:

```text
4 часа
оценка 6
```

````python
print("Старая модель:")
ask_ai(4, 6)
````

Теперь добавим новые примеры:

```text
[4, 6] → 0
[4, 5] → 0
[5, 5] → 0
```

И создадим **новую модель**.

````python
X_new = X + [
    [4, 6],
    [4, 5],
    [5, 5]
]

y_new = y + [
    0,
    0,
    0
]

model_new = MLPClassifier(
    hidden_layer_sizes=(5,),
    solver="lbfgs",
    max_iter=2000,
    random_state=42
)

model_new.fit(X_new, y_new)

print("Новая модель обучена!")
````

````python
old_answer = model.predict([[4, 6]])[0]
new_answer = model_new.predict([[4, 6]])[0]

print("Старая модель:", old_answer)
print("Новая модель: ", new_answer)
````

## Что произошло?

Мы **не меняли программу нейросети**.

Мы изменили только данные.

Но поведение модели изменилось.

Запомните:

```text
ДАННЫЕ → ОБУЧЕНИЕ → ПОВЕДЕНИЕ ИИ
```

Поэтому качество и правильность данных очень важны.

# Часть 11. Можно ли плохо обучить ИИ?

Да.

Давайте специально дадим модели неправильные ответы.

Будем говорить:

- плохие оценки → `1`;
- хорошие оценки → `0`.

Это специально испорченные данные.

````python
wrong_y = [
    1,
    1,
    1,
    1,
    0,
    0,
    0,
    0
]

bad_model = MLPClassifier(
    hidden_layer_sizes=(5,),
    solver="lbfgs",
    max_iter=2000,
    random_state=42
)

bad_model.fit(X, wrong_y)

print("Модель обучена на неправильных данных.")
````

````python
tests = [
    [1, 2],
    [8, 10]
]

for person in tests:
    good_answer = model.predict([person])[0]
    bad_answer = bad_model.predict([person])[0]

    print(
        person,
        "→ правильная модель:",
        good_answer,
        "| испорченная модель:",
        bad_answer
    )
````

## Вывод

ИИ не знает, где правда.

Он изучает те примеры, которые ему дали.

Поэтому:

> **Плохие данные могут научить ИИ неправильному поведению.**

# Часть 12. Мини-проект: ИИ определяет фрукт 🍎🍊

Теперь попробуйте создать вторую модель.

У нас два признака:

1. вес фрукта в граммах;
2. диаметр в сантиметрах.

Ответ:

- `0` — яблоко;
- `1` — апельсин.

Данные уже подготовлены.

````python
fruit_X = [
    [120, 6],
    [140, 7],
    [150, 7],
    [160, 8],

    [180, 8],
    [200, 9],
    [220, 10],
    [240, 10]
]

fruit_y = [
    0,
    0,
    0,
    0,
    1,
    1,
    1,
    1
]

print("Данные готовы!")
````

## Задание 1

Создайте нейронную сеть самостоятельно.

Подсказка:

```python
fruit_model = MLPClassifier(
    hidden_layer_sizes=(...),
    solver="lbfgs",
    max_iter=2000,
    random_state=42
)
```

Вы можете выбрать, например, `4`, `5` или `8` нейронов.

````python
fruit_model = MLPClassifier(
    hidden_layer_sizes=(5,),
    solver="lbfgs",
    max_iter=2000,
    random_state=42
)
````

## Задание 2

Обучите модель.

Нужна всего одна команда:

```python
fruit_model.fit(...)
```

````python
fruit_model.fit(fruit_X, fruit_y)

print("ИИ научился различать фрукты!")
````

## Задание 3

Проверьте новый фрукт:

```text
вес = 210 г
диаметр = 9 см
```

````python
fruit = [[210, 9]]

answer = fruit_model.predict(fruit)[0]

if answer == 0:
    print("🍎 ИИ считает: яблоко")
else:
    print("🍊 ИИ считает: апельсин")
````

## Задание 4

Проверьте ещё несколько фруктов:

```text
130 г, 6 см
170 г, 8 см
230 г, 10 см
190 г, 7 см
```

Перед запуском сначала **сами предположите ответ**.

````python
fruits = [
    [130, 6],
    [170, 8],
    [230, 10],
    [190, 7]
]

for fruit in fruits:
    answer = fruit_model.predict([fruit])[0]

    if answer == 0:
        name = "🍎 яблоко"
    else:
        name = "🍊 апельсин"

    print(fruit, "→", name)
````
